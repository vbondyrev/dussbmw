## Save processed and transformed tables to the Gold Zone for reporting needs

```python

table_name = []

for i in dbutils.fs.ls('mnt/silver/public'):
    table_name.append(i.name.split('/')[0])

table_name
```

```python
for name in table_name:
    path = '/mnt/silver/public/' + name
    df = spark.read.format('delta').load(path)

    output_path = '/mnt/gold/' + name + '/'
    df.write.format('delta').mode('overwrite').save(output_path)
    print(f"Delta table saved in Gold zone as: {output_path}")
```
